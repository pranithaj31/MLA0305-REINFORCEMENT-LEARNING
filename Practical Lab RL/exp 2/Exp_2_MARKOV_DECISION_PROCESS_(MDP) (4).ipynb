```python
# ================================================================
# EXPERIMENT 2
# MARKOV DECISION PROCESS (MDP)
# ================================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

print("=" * 70)
print("EXPERIMENT 2")
print("MARKOV DECISION PROCESS (MDP)")
print("=" * 70)

# ================================================================
# AIM
# ================================================================

print("\nAIM")
print("-" * 70)

print(
    "To implement a Markov Decision Process (MDP) by defining "
    "states, actions, transitions, rewards and discount factor."
)

# ================================================================
# THEORY
# ================================================================

print("\nTHEORY")
print("-" * 70)

print("An MDP is represented as:")
print("MDP = (S, A, P, R, Gamma)")

print("\nWhere:")
print("S     = Set of States")
print("A     = Set of Actions")
print("P     = Transition Probability")
print("R     = Reward Function")
print("Gamma = Discount Factor")

# ================================================================
# STEP 1: DEFINE STATES
# ================================================================

states = [
    "S0",
    "S1",
    "S2",
    "S3",
    "Goal"
]

# ================================================================
# STEP 2: DEFINE ACTIONS
# ================================================================

actions = [
    "Left",
    "Right"
]

# ================================================================
# STEP 3: DISCOUNT FACTOR
# ================================================================

gamma = 0.90

print("\nDiscount Factor Gamma =", gamma)

# ================================================================
# STEP 4: DEFINE TRANSITIONS
# ================================================================

# Transition format:
#
# Current State
#      |
#    Action
#      |
#      v
# Next State + Reward

transition_table = {

    ("S0", "Right"): ("S1", -1),
    ("S0", "Left"): ("S0", -1),

    ("S1", "Right"): ("S2", -1),
    ("S1", "Left"): ("S0", -1),

    ("S2", "Right"): ("S3", -1),
    ("S2", "Left"): ("S1", -1),

    ("S3", "Right"): ("Goal", 10),
    ("S3", "Left"): ("S2", -1)
}

# ================================================================
# STEP 5: DISPLAY MDP COMPONENTS
# ================================================================

print("\n")
print("=" * 70)
print("MDP COMPONENTS")
print("=" * 70)

print("\nStates:")
print(states)

print("\nActions:")
print(actions)

print("\nDiscount Factor:")
print(gamma)

# ================================================================
# STEP 6: CREATE TRANSITION SUMMARY TABLE
# ================================================================

transition_data = []

for key, value in transition_table.items():

    current_state = key[0]
    action = key[1]

    next_state = value[0]
    reward = value[1]

    transition_data.append([
        current_state,
        action,
        next_state,
        reward
    ])

transition_df = pd.DataFrame(
    transition_data,
    columns=[
        "Current State",
        "Action",
        "Next State",
        "Reward"
    ]
)

print("\n")
print("=" * 70)
print("TRANSITION TABLE")
print("=" * 70)

print(
    transition_df.to_string(index=False)
)

# ================================================================
# STEP 7: SIMULATE AN AGENT
# ================================================================

print("\n")
print("=" * 70)
print("AGENT SIMULATION")
print("=" * 70)

current_state = "S0"

episode = []

total_reward = 0

discounted_reward = 0

step = 0

# Agent follows the optimal path:
# S0 -> S1 -> S2 -> S3 -> Goal

while current_state != "Goal":

    if current_state == "S0":
        action = "Right"

    elif current_state == "S1":
        action = "Right"

    elif current_state == "S2":
        action = "Right"

    elif current_state == "S3":
        action = "Right"

    next_state, reward = transition_table[
        (current_state, action)
    ]

    step += 1

    total_reward += reward

    discounted_reward += (
        (gamma ** (step - 1)) * reward
    )

    episode.append([
        step,
        current_state,
        action,
        next_state,
        reward
    ])

    print(
        "Step",
        step,
        ":",
        current_state,
        "--",
        action,
        "-->",
        next_state,
        "Reward =",
        reward
    )

    current_state = next_state

# ================================================================
# STEP 8: EPISODE TABLE
# ================================================================

episode_df = pd.DataFrame(
    episode,
    columns=[
        "Step",
        "Current State",
        "Action",
        "Next State",
        "Reward"
    ]
)

print("\n")
print("=" * 70)
print("EPISODE SUMMARY")
print("=" * 70)

print(
    episode_df.to_string(index=False)
)

# ================================================================
# STEP 9: CALCULATE REWARD
# ================================================================

print("\n")
print("=" * 70)
print("REWARD ANALYSIS")
print("=" * 70)

print(
    "Total Reward      :",
    total_reward
)

print(
    "Discounted Reward :",
    round(discounted_reward, 3)
)

print(
    "Number of Steps   :",
    step
)

print(
    "Final State       :",
    current_state
)

# ================================================================
# STEP 10: SUMMARY TABLE
# ================================================================

summary_table = pd.DataFrame({

    "Parameter": [
        "Number of States",
        "Number of Actions",
        "Discount Factor",
        "Number of Steps",
        "Total Reward",
        "Discounted Reward",
        "Initial State",
        "Final State"
    ],

    "Value": [
        len(states),
        len(actions),
        gamma,
        step,
        total_reward,
        round(discounted_reward, 3),
        "S0",
        current_state
    ]
})

print("\n")
print("=" * 70)
print("SUMMARY TABLE")
print("=" * 70)

print(
    summary_table.to_string(index=False)
)

# ================================================================
# STEP 11: SAVE RESULTS
# ================================================================

transition_df.to_csv(
    "MDP_Transition_Table.csv",
    index=False
)

episode_df.to_csv(
    "MDP_Episode_Results.csv",
    index=False
)

summary_table.to_csv(
    "MDP_Summary.csv",
    index=False
)

# ================================================================
# VISUALIZATION 1
# REWARD PER STEP
# ================================================================

plt.figure(figsize=(9, 5))

plt.bar(
    episode_df["Step"],
    episode_df["Reward"],
    color=[
        "skyblue",
        "lightgreen",
        "gold",
        "salmon"
    ]
)

plt.title(
    "MDP Reward per Step",
    fontsize=14
)

plt.xlabel("Step")

plt.ylabel("Reward")

plt.grid(
    axis="y",
    alpha=0.3
)

plt.show()

# ================================================================
# VISUALIZATION 2
# CUMULATIVE REWARD
# ================================================================

cumulative_reward = np.cumsum(
    episode_df["Reward"]
)

plt.figure(figsize=(9, 5))

plt.plot(
    episode_df["Step"],
    cumulative_reward,
    marker="o",
    linewidth=3,
    color="purple"
)

plt.title(
    "Cumulative Reward",
    fontsize=14
)

plt.xlabel("Step")

plt.ylabel("Cumulative Reward")

plt.grid(True, alpha=0.3)

plt.show()

# ================================================================
# VISUALIZATION 3
# STATE TRANSITION PATH
# ================================================================

state_numbers = np.arange(
    len(episode_df) + 1
)

state_path = [
    "S0",
    "S1",
    "S2",
    "S3",
    "Goal"
]

plt.figure(figsize=(10, 5))

plt.plot(
    state_numbers,
    np.arange(len(state_numbers)),
    marker="o",
    linewidth=3,
    color="green"
)

plt.xticks(
    state_numbers,
    state_path
)

plt.title(
    "MDP State Transition Path",
    fontsize=14
)

plt.xlabel("State")

plt.ylabel("Transition Order")

plt.grid(True, alpha=0.3)

plt.show()

# ================================================================
# FINAL RESULT
# ================================================================

print("\n")
print("=" * 70)
print("FINAL RESULT")
print("=" * 70)

print(
    "The Markov Decision Process was successfully implemented."
)

print(
    "The agent started from S0 and reached the Goal state."
)

print(
    "Total Reward      :",
    total_reward
)

print(
    "Discounted Reward :",
    round(discounted_reward, 3)
)

# ================================================================
# CONCLUSION
# ================================================================

print("\n")
print("=" * 70)
print("CONCLUSION")
print("=" * 70)

print(
    "The MDP successfully demonstrated sequential decision-making."
)

print(
    "The agent selected actions, transitioned between states,"
)

print(
    "received rewards, and finally reached the Goal state."
)

print(
    "Thus, the basic components of an MDP were successfully"
)

print(
    "implemented and verified using Python."
)

print("\nExperiment 2 Completed Successfully.")
```
